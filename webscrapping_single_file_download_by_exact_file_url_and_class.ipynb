```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

# Step 1: Request the arXiv homepage (though ideally you should target a specific paper page)
url = 'https://arxiv.org/abs/2504.18555'
response = requests.get(url)

# Step 2: Parse HTML if successful
if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'html.parser')

    # Step 3: Find the PDF download link
    link_tag = soup.find('a', class_='abs-button download-pdf')
    if link_tag and 'href' in link_tag.attrs:
        relative_link = link_tag['href']
        full_url = urljoin(url, relative_link)

        print('Document link found:', full_url)

        # Optional: Download the PDF
        pdf_response = requests.get(full_url)
        if pdf_response.status_code == 200:
            with open('downloaded_paper.pdf', 'wb') as f:
                f.write(pdf_response.content)
            print("PDF downloaded successfully.")
        else:
            print("Failed to download the PDF.")
    else:
        print('PDF link not found.')
else:
    print('Failed to retrieve the webpage.')
```

```python
import requests
from bs4 import BeautifulSoup
import urllib.parse
import re
import time

# Base URL for the arXiv page
base_url = 'https://arxiv.org'

# URL of the page to scrape
url = 'https://arxiv.org/list/math-ph/new'

try:
    # Send a GET request to fetch the page content
    response = requests.get(url)
    response.raise_for_status()  # Raise an exception for bad status codes

    # Parse the page content with BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # Find all links that point to PDF files (href contains '/pdf/')
    document_links = soup.find_all('a', href=re.compile(r'^/pdf/.*'))

    if not document_links:
        print("No PDF links found on the page. Check the page structure or regex pattern.")
        exit()

    # Loop through each link and download the corresponding PDF
    for i, link in enumerate(document_links):
        # Construct the full URL for the PDF
        document_url = urllib.parse.urljoin(base_url, link['href']) + '.pdf'  # Append .pdf to ensure PDF download

        try:
            # Fetch the PDF
            document_response = requests.get(document_url)
            document_response.raise_for_status()  # Raise an exception for bad status codes

            # Extract arXiv ID from the URL for the filename
            arxiv_id = link['href'].split('/')[-1]  # e.g., 1234.5678v1
            file_name = f'arxiv_{arxiv_id}.pdf'

            # Save the PDF
            with open(file_name, 'wb') as file:
                file.write(document_response.content)
            print(f'Document {i+1} downloaded successfully as {file_name}.')

            # Add a delay to avoid overwhelming the server
            time.sleep(1)

        except requests.RequestException as e:
            print(f'Failed to download document {i+1} from {document_url}. Error: {e}')

except requests.RequestException as e:
    print(f'Failed to fetch the webpage {url}. Error: {e}')
except Exception as e:
    print(f'An unexpected error occurred: {e}')
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("homayoonkhadivi/nhanes-datasets")

print("Path to dataset files:", path)
```

```python
# prompt: how can i get this data from path and make dataframe

import pandas as pd
import os

# Assuming 'path' variable holds the directory containing the dataset files.
# Replace with the actual path if needed.

# Iterate through all files in the directory
for filename in os.listdir(path):
    filepath = os.path.join(path, filename)

    # Check if it's a CSV file (adjust file type if needed)
    if filename.endswith(".csv"):
        try:
            # Create a DataFrame for each CSV file
            df = pd.read_csv(filepath)
            print(f"DataFrame created for {filename}:\n{df.head()}")  # Print first few rows

            # You can perform further operations on the DataFrame 'df' here
            # For example, save to a new file, analyze data, etc.
        except Exception as e:
            print(f"Error processing {filename}: {e}")
    else:
        print(f"Skipping non-CSV file: {filename}")
```
